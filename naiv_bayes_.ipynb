```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes  import GaussianNB
```

```python
%cd /content/drive/MyDrive/Course Material/Supervised Machine Learning/Bayes Theorem and Naive Bayes Classifier
```

```python
data  =  pd.read_csv('titanic_train.csv')
data.head()
```

```python
data.drop(['PassengerId','Name','Cabin','Ticket'] ,  inplace =  True , axis =  1)
data
```

```python
data.isnull().any()
```

```python
data.fillna(value  = { "Age": data['Age'].mean()},inplace  =  True)
```

```python
data.isnull().any()
```

```python
data.dropna(inplace =  True)
data.isnull().any()
```

```python
data
```

```python
dummy = pd.get_dummies(data['Sex'])
dummy_1  = pd.get_dummies(data['Embarked'])
```

```python
new_data  = pd.concat([data,dummy,dummy_1],axis = 1)
new_data
```

```python
new_data = new_data.drop(['Sex','Embarked','female'],axis = 1 )
new_data
```

```python
x =  new_data.drop('Survived',axis = 1).values
y = new_data['Survived'].values
```

```python

from sklearn.preprocessing import StandardScaler
x_train,x_test,y_train,y_test  = train_test_split(x,y,test_size = 0.2,random_state  = 42)
sc  =  StandardScaler()
sc.fit_transform(x_train)
sc.transform(x_test)

naiv = GaussianNB()
naiv.fit(x_train,y_train)
```
